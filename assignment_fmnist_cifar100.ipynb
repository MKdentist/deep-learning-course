```python
import tensorflow as tf
```

```python
d= tf.keras.datasets.cifar100
```

```python
d
```

```python
(training_images, training_labels), (test_images, test_labels) = d.load_data()
```

```python
len(training_images)
```

```python
len(test_images)
```

```python
test_labels
```

```python
import numpy as np
import matplotlib.pyplot as plt

index = 5

np.set_printoptions(linewidth=320)
print(f'\nIMAGE PIXEL ARRAY:\n {training_images[index]}')
print(f'LABEL: {training_labels[index]}')
plt.imshow(training_images[index])
```

```python
training_images  = training_images / 255.0
test_images = test_images / 255.0
```

```python
model= tf.keras.models.Sequential([tf.keras.layers.Flatten(),
                                    tf.keras.layers.Dense(728, activation=tf.nn.relu),
                                     tf.keras.layers.Dense(324, activation=tf.nn.relu),
                                     tf.keras.layers.Dense(164, activation=tf.nn.relu),
                                    tf.keras.layers.Dense(100, activation=tf.nn.softmax)])
```

```python
model.compile(optimizer = tf.optimizers.Adam(),
              loss = 'sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.fit(training_images, training_labels, epochs=50)
```

```python
model.evaluate(test_images, test_labels)
```

```python
classifications = model.predict(test_images)
# in classifications predict is saved for test images
print(classifications[100])
```

```python
from google.colab import files
from PIL import Image, ImageOps
import numpy as np
import matplotlib.pyplot as plt

class_names = [
    "apple", "aquarium_fish", "baby", "bear", "beaver", "bed", "bee", "beetle",
    "bicycle", "bottle", "bowl", "boy", "bridge", "bus", "butterfly", "camel",
    "can", "castle", "caterpillar", "cattle", "chair", "chimpanzee", "clock",
    "cloud", "cockroach", "computer_keyboard", "couch", "crab", "crocodile",
    "cup", "dinosaur", "dolphin", "elephant", "flatfish", "forest", "fox",
    "girl", "hamster", "house", "kangaroo", "lamp", "lawn_mower", "leopard",
    "lion", "lizard", "lobster", "man", "maple_tree", "motorcycle", "mountain",
    "mouse", "mushroom", "oak_tree", "orange", "orchid", "otter", "palm_tree",
    "pear", "pickup_truck", "pine_tree", "plain", "plate", "poppy", "porcupine",
    "possum", "rabbit", "raccoon", "ray", "road", "rocket", "rose", "sea",
    "seal", "shark", "shrew", "skunk", "skyscraper", "snail", "snake", "spider",
    "squirrel", "streetcar", "sunflower", "sweet_pepper", "table", "tank",
    "telephone", "television", "tiger", "tractor", "train", "trout", "tulip",
    "turtle", "wardrobe", "whale", "willow_tree", "wolf", "woman", "worm"]
```

```python
uploaded = files.upload()
for file_name in uploaded.keys():
    img = Image.open(file_name).convert("RGB") # Convert to RGB to ensure 3 channels
    img = img.resize((32, 32)) # Resize to 32x32 to match training data
    # img = ImageOps.invert(img) # Inverting might not be necessary for color images or CIFAR-100
    img_array = np.array(img) / 255.0
    img_input = np.expand_dims(img_array, axis=0)

    prediction = model.predict(img_input)
    predicted_index = np.argmax(prediction[0])
    confidence = np.max(prediction[0]) * 100
    plt.imshow(img_array)
    plt.axis("off")
    plt.title(f"Prediction: {class_names[predicted_index]} ({confidence:.2f}% confidence)")
    plt.show()

    print("Uploaded file:", file_name)
    print("Predicted class number:", predicted_index)
    print("Predicted class name:", class_names[predicted_index])
    print(f"Confidence: {confidence:.2f}%")
    print("All class probabilities:")
    for name, prob in zip(class_names, prediction[0]):
        print(f"{name}: {prob*100:.2f}%")
```
